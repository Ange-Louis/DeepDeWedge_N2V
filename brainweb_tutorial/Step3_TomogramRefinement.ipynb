
## Refining the BrainWeb Data

The `ddw refine-tomogram` command produces the final denoised and missing-wedge-corrected tomogram(s) using a model checkpoint produced during model fitting. You determine which checkpoint to use by setting the `model_checkpoint_file` parameter in `brainweb_config.yaml`. You can find checkpoint files (`.ckpt` extension) in `./brainweb_project/logs/version_0/checkpoints`. As in the original tutorial, this directory contains three sub-directories:

- `epoch/`: Checkpoints periodically created after a fixed number of epochs
- `fitting_loss/`: Checkpoints created when the model achieved lowest fitting losses 
- `val_loss/`: Checkpoints created when the model achieved the lowest validation losses

The frequency and criteria for the creation of these checkpoints are arguments of `ddw fit-model`; see the output of `ddw fit-model --help` for details.

Provided that there was no overfitting, the checkpoint corresponding to the highest epoch should give good results, but you can also try the ones with the lowest validation or fitting losses.

**Note:** unlike the original tutorial, there is no pre-fitted checkpoint shipped with the BrainWeb data — `model_checkpoint_file` in `brainweb_config.yaml` must point to a checkpoint produced by your own `ddw fit-model` run in Step 2, e.g. `./brainweb_project/logs/chemin/de/mon/meilleur/model.ckpt`. Adjust this path before running the cell below.
